# 1. Importar las librerias


```python
import requests
import pandas as pd
```

# 2. Consultar la API

```python
url = "https://rickandmortyapi.com/api/character"

respuesta = requests.get(url)

datos = respuesta.json()

print(datos['info'])
```

# 3. Mostrar los personajes

```python
personajes = pd.DataFrame(datos['results'])

display(personajes[['id', 'name', 'status', 'species', 'gender']].head(10))
```

# 4. Recorrer todas las páginas de la API

```python
personajes = []
url = "https://rickandmortyapi.com/api/character"

while url:
    respuesta = requests.get(url, timeout=30)

    if respuesta.status_code != 200:
        print("Error al consultar la API:", respuesta.status_code)
        break

    datos = respuesta.json()

    personajes.extend(datos['results'])
    url = datos['info']['next']

df = pd.DataFrame(personajes)

display(df.head())
print("Total de personajes:", len(df))
```

# 5. Guardar los datos en csv

```python


df.to_csv('data/personajes.csv', index=False, encoding='utf-8-sig')

print("Archivo guardado correctamente")
```
