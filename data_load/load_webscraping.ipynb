## 1. Importar las librerias

```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
```

## 2. Conectarse a la página web

```python
url = "https://books.toscrape.com/"

respuesta = requests.get(url)

soup = BeautifulSoup(respuesta.text, 'html.parser')

print("Página conectada correctamente")
```

## 3. Extraer informacion de los libros

```python
libros = []

for libro in soup.select('.product_pod'):
    titulo = libro.h3.a['title']
    precio = libro.select_one('.price_color').text
    disponibilidad = libro.select_one('.availability').text.strip()

    libros.append({
        'titulo': titulo,
        'precio': precio,
        'disponibilidad': disponibilidad
    })

df = pd.DataFrame(libros)

display(df.head(10))
```

## 4. Guardar los datos en CSV

```python
df.to_csv('data/libros.csv', index=False, encoding='utf-8-sig')

print("Archivo CSV guardado correctamente")
```

## 5. Mostrar información del conjunto de Datos

```python
print("Total de libros:", len(df))
display(df.describe(include='all'))
```
