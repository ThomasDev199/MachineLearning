## 4. Rellenar los valores nulos con la media de la columna


```python
df['precio_unitario'] = df['precio_unitario'].fillna(df['precio_unitario'].mean())
df.head()
```

```python
# 5. Verificar si hay valores duplicados
df.duplicated().sum()

# 6. Eliminar valores duplicados
# df.drop_duplicades()
```

## 2. Codifique variables categóricas usando LabelEncoder y One-Hot Encoding

```python
from sklearn.preprocessing import LabelEncoder

df['categoria'] = LabelEncoder().fit_transform(df['categoria'])

df = pd.get_dummies(df, columns=['ciudad', 'producto'], dtype=int)

display(df.head())
```

## 3. Aplique normalización y estandarizacion.

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

columnas = ['cantidad', 'precio_unitario', 'total_venta']

df[columnas] = MinMaxScaler().fit_transform(df[columnas])

df[columnas] = StandardScaler().fit_transform(df[columnas])

display(df.head())
```

## 4. Balancee la variable objetvo usando SMOTE.

```python
from imblearn.over_sampling import SMOTE

X = df.drop(columns=['categoria', 'fecha'])
y = df['categoria']

X, y = SMOTE(random_state=42, k_neighbors=1).fit_resample(X, y)

print(y.value_counts())
```

## 5. Muestre el conjunto de datos final listo para entrenar un modelo de clasificación.

```python
df_final = X.copy()
df_final['categoria'] = y

display(df_final.head())

print("Tamaño del conjunto:", df_final.shape)
```
